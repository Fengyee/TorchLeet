# Problem: Implement Linear Regression

### Problem Statement
Your task is to implement a **Linear Regression** model using PyTorch. The model predicts a continuous target $y$ from a single input feature $x$ as $\hat{y} = wx + b$, and is trained with mean squared error and gradient descent.

### Requirements
1. **Model Definition**:
   - Implement a class `LinearRegressionModel` (an `nn.Module`) with:
     - A single linear layer, stored as `self.linear`, mapping 1 input feature to 1 output (a weight **and** a bias).
2. **Forward Method**:
   - Implement the `forward` method: input of shape `(N, 1)` → predictions of shape `(N, 1)`.

The training loop is provided. The last cell compares what gradient descent learned against the closed-form least-squares solution — the two should agree closely.

```python
import torch
import torch.nn as nn
import torch.optim as optim
```

```python
# Generate synthetic data
torch.manual_seed(42)
X = torch.rand(100, 1) * 10  # 100 data points between 0 and 10
y = 2 * X + 3 + torch.randn(100, 1)  # Linear relationship with noise

# Define the Linear Regression Model
class LinearRegressionModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.linear = nn.Linear(1, 1)  # y_hat = w * x + b: one input feature, one output

    def forward(self, x):
        return self.linear(x)

# Initialize the model, loss function, and optimizer
model = LinearRegressionModel()
criterion = nn.MSELoss()
optimizer = optim.SGD(model.parameters(), lr=0.01)

# Training loop
epochs = 1000
for epoch in range(epochs):
    # Forward pass
    predictions = model(X)
    loss = criterion(predictions, y)

    # Backward pass and optimization
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    # Log progress every 100 epochs
    if (epoch + 1) % 100 == 0:
        print(f"Epoch [{epoch + 1}/{epochs}], Loss: {loss.item():.4f}")
```

```python
# Display the learned parameters
[w, b] = model.linear.parameters()
print(f"Learned weight: {w.item():.4f}, Learned bias: {b.item():.4f}")

# Sanity check: ordinary least squares has a closed-form solution. Solve it on the
# design matrix [x, 1] and compare with what gradient descent found.
A = torch.cat([X, torch.ones_like(X)], dim=1)
w_ols, b_ols = torch.linalg.lstsq(A, y).solution.flatten().tolist()
print(f"Closed-form OLS: weight {w_ols:.4f}, bias {b_ols:.4f}")

# Testing on new data
X_test = torch.tensor([[4.0], [7.0]])
with torch.no_grad():
    predictions = model(X_test)
    print(f"Predictions for {X_test.tolist()}: {predictions.tolist()}")
```
